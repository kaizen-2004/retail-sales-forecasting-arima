# Sales Forecasting - Automation & Deployment

This notebook tests the automated pipeline and generates final documentation.

````python
import sys
from pathlib import Path

# Add src to path
sys.path.insert(0, str(Path('..').resolve()))

from src.pipeline import run_pipeline
````

## 1. Run Automated Pipeline

````python
# Run the complete pipeline
print("Running automated pipeline...")
print("="*60)

results = run_pipeline()

print("\n" + "="*60)
print("PIPELINE RESULTS")
print("="*60)
print(f"Status: {results['status']}")
if results['status'] == 'success':
    print(f"MAPE: {results['metrics']['mape']:.2f}%")
    print(f"Report: {results['report_path']}")
else:
    print(f"Error: {results.get('error', 'Unknown')}")
````

## 2. Verify Outputs

````python
# Check generated files
import os

files_to_check = [
    'data/processed/daily_sales_latest.csv',
    'data/processed/forecast_latest.csv',
    'reports/latest_report.txt',
]

print("Generated files:")
print("-"*40)
for f in files_to_check:
    path = Path('..') / f
    if path.exists():
        size = os.path.getsize(path)
        print(f"✓ {f} ({size:,} bytes)")
    else:
        print(f"✗ {f} (not found)")
````

## 3. View Report

````python
# Display the report
report_path = Path('../reports/latest_report.txt')
if report_path.exists():
    with open(report_path) as f:
        print(f.read())
else:
    print("Report not found. Run the pipeline first.")
````
